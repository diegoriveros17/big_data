## 17. Proyecto integrador

Construir un programa que simule una conversación sencilla entre una persona y la computadora. El programa debe reconocer algunas palabras o frases preestablecidas y responder según un conjunto de reglas heurísticas implementadas con condiciones. No es necesario usar inteligencia artificial ni conectarse a Internet: alcanza con analizar el texto ingresado y buscar coincidencias.

### Requisitos mínimos

- Guardar en un arreglo una colección de palabras o frases que el chatbot pueda reconocer, por ejemplo saludos, preguntas sobre su estado, pedidos de ayuda y una despedida.
- Solicitar mensajes de entrada y responder con reglas heurísticas usando `if`, `elif` y `else`.
- Considerar que una misma palabra puede aparecer dentro de frases diferentes y definir qué respuesta tiene prioridad cuando coinciden varias reglas.
- Mantener la conversación con un bucle hasta que la persona escriba una palabra de salida, como `salir` o `adiós`.
- Organizar la lógica en una o más funciones con nombres claros.
- Validar entradas vacías o con un formato no esperado.
- Anticipar posibles errores, como datos inválidos o problemas al leer una entrada, y capturarlos con `try` / `except` de forma específica. El programa no debe terminar abruptamente ante esos casos.
- Mostrar mensajes claros para las respuestas reconocidas y una respuesta general cuando no se encuentre ninguna coincidencia.

### Preguntas de revisión

- ¿Qué palabras o frases reconoce el chatbot y cómo están organizadas en el arreglo?
- ¿Qué regla se aplica si un mensaje coincide con más de una categoría?
- ¿Qué sucede si la entrada está vacía, es desconocida o produce un error?
- ¿Dónde empieza y termina cada responsabilidad?
- ¿Qué casos de prueba demuestran que funcionan las respuestas y la captura de errores?
- ¿Qué regla o respuesta agregarías para que la conversación sea más completa?

```python
PALABRAS = [
    ["hola", "buenas", "buen dia", "buenas tardes"],
    ["como estas", "como te va", "que tal"],          
    ["ayuda", "necesito", "no entiendo"],                 
    ["adios", "chau", "hasta luego", "salir"]         
]

def respuestas(mensaje: str) -> tuple[str, bool]:
    mensaje = mensaje.lower().strip()
    
    # Validar entrada vacía
    if not mensaje:
        return "Por favor, escribe algo para poder ayudarte.", False
    
    # Despedida 
    for palabra in PALABRAS[3]:
        if palabra in mensaje:
            return "¡Hasta luego! Que tengas un excelente día.", True
            
    # Pedido de ayuda
    for palabra in PALABRAS[2]:
        if palabra in mensaje:
            return "Puedo ayudarte respondiendo preguntas simples. Escribe 'hola', 'cómo estás' o 'chau'.", False
            
    # Estado
    for palabra in PALABRAS[1]:
        if palabra in mensaje:
            return "Bien! Gracias por preguntar.", False
            
    # Saludos
    for palabra in PALABRAS[0]:
        if palabra in mensaje:
            return "¡Hola! ¿En qué puedo ayudarte hoy?", False
            
    # Respuesta por defecto cuando no hay coincidencias
    return "Lo siento, no entendí tu mensaje. ¿Podrías escribirlo de otra forma?", False


def chatbot():
    print("Escribe 'salir' para terminar\n")
    
    salir = False
    
    while not salir:
        try:
            input_user = input("Tú: ")
            respuesta, salir = respuestas(input_user)
            print(f"Chatbot: {respuesta}\n")
        except Exception as e:
            print(f"\nChatbot: Ocurrió un error inesperado ({e}). Finalizando programa.\n")
            break  

chatbot()
```
