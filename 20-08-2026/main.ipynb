# Python
Python es un lenguaje de programación de alto nivel, tipado dinamico, esta escrito en C, y su implementación oficial es CPython. Fue creado por Guido Van Rossum en el 1991.  

La programación consiste en expresar una solución como una secuencia clara de pasos. Python nos ayuda a escribir esos pasos de una forma legible.

## 0. Entornos virtuales de Python

Un **entorno virtual** es una carpeta que contiene un entorno de Python separado del resto de la computadora. En él se pueden instalar paquetes y herramientas para un proyecto sin modificar la instalación general de Python ni afectar a otros proyectos.

### ¿Por qué usamos un entorno virtual?

Cada proyecto puede necesitar paquetes diferentes o versiones distintas de un mismo paquete. Si todo se instala en el Python global, los proyectos pueden entrar en conflicto. El entorno virtual permite:

- mantener aisladas las dependencias de cada proyecto;
- evitar modificar el Python del sistema;
- trabajar con versiones específicas de paquetes;
- reproducir el entorno en otra computadora;
- eliminar y volver a crear el entorno sin perder el código del proyecto.

El entorno virtual no reemplaza al proyecto ni contiene nuestro código fuente: contiene una instalación aislada de Python y los paquetes que el proyecto necesita. Por convención, suele llamarse `.venv` y no se sube al control de versiones. Es un entorno descartable: si se rompe, se puede recrear.

> **Idea clave:** el código del proyecto y el entorno virtual son cosas diferentes. El código se conserva; el entorno se puede reconstruir.

### Antes de crearlo

Abrir una terminal ubicada en la carpeta del proyecto. Se puede crear el entorno con la herramienta estándar de Python (`venv`) o con `uv`. Para un mismo proyecto se elige una de las dos formas; no es necesario crear dos entornos.

## Crear el entorno sin `uv`

Python incluye el módulo `venv`, que permite crear entornos virtuales sin instalar una herramienta adicional.

En macOS, Linux y también en Windows si el comando `python` está disponible:

```text
python -m venv .venv
```

En algunos sistemas el intérprete se invoca como `python3` o `py`. En ese caso se utiliza el comando equivalente:

```text
python3 -m venv .venv
py -m venv .venv
```

El último argumento, `.venv`, es el nombre de la carpeta que se va a crear. Se puede elegir otro nombre, pero `.venv`,  `venv` o `env` es la convención más habitual.

### Activar el entorno

Crear el entorno y activarlo son pasos diferentes. La activación hace que los comandos `python` y `pip` de la terminal apunten al entorno del proyecto.

**macOS o Linux — Bash/Zsh**

```text
source .venv/bin/activate
```

**Windows — PowerShell**

```text
.venv\Scripts\activate
```

**Windows — CMD**

```text
.venv\Scripts\activate.bat  o .venv\Scripts\activate
```

Cuando está activo, normalmente aparece `(.venv)` al comienzo de la línea de la terminal. Para instalar un paquete en este entorno se puede usar `python -m pip install nombre-del-paquete`. Para salir del entorno se utiliza `deactivate`.

## Crear el entorno con `uv`

`uv` es una herramienta rápida para trabajar con versiones de Python, entornos virtuales y dependencias de proyectos. Debe estar instalado antes de utilizar estos comandos.

Desde la carpeta del proyecto, crear el entorno predeterminado `.venv` con:

```text
uv venv .venv
```

También se puede solicitar una versión concreta de Python, siempre que esté disponible o `uv` pueda instalarla:

```text
uv venv .venv --python 3.12
```

El entorno creado con `uv` se activa con los mismos comandos que cualquier otro entorno virtual: `source .venv/bin/activate` en macOS/Linux, `.venv\Scripts\activate` en PowerShell o `.venv\Scripts\activate.bat` en CMD.

Una vez creado, `uv` puede instalar paquetes en ese entorno:

```text
uv pip install nombre-del-paquete
```

En proyectos administrados por `uv`, también se puede utilizar `uv init` para iniciar el proyecto, `uv add nombre-del-paquete` para declarar una dependencia y `uv run` para ejecutar un comando dentro del entorno del proyecto. En ese flujo, `uv` mantiene el entorno y las dependencias sincronizados.

```text
uv init
uv add nombre-del-paquete
uv run python archivo.py
```

Con `uv run` no es obligatorio activar manualmente el entorno para ejecutar el comando. Activarlo puede ser útil para que el editor y la terminal lo reconozcan de forma explícita.

## Comprobar qué Python estamos utilizando

Cuando haya dudas, se puede consultar la versión y la ubicación del intérprete que está ejecutándose:

```text
python --version
python -c "import sys; print(sys.executable)"
```

La ruta mostrada debería apuntar a la carpeta `.venv` cuando el entorno esté activo o cuando el comando se ejecute dentro del entorno administrado por `uv`.

### Comparación rápida

| Forma | Creación | Instalación habitual |
| --- | --- | --- |
| Python estándar | `python -m venv .venv` | `python -m pip install paquete` |
| Con `uv` | `uv venv` | `uv pip install paquete` o `uv add paquete` en un proyecto `uv` |

### Reglas importantes

- No guardar la carpeta `.venv` en el repositorio: se puede recrear.
- No mezclar paquetes del sistema con los del proyecto sin una razón clara.
- Activar el entorno antes de instalar paquetes cuando se trabaja con `venv` y `pip`.
- Con `uv`, recordar que `uv run` ejecuta el comando dentro del entorno del proyecto.
- Registrar las dependencias del proyecto para que otra persona pueda reconstruir el entorno.

## 1. La sintaxis de Python

La **sintaxis** es el conjunto de reglas que indica cómo se deben escribir las instrucciones. Python busca que el código sea legible, pero sigue siendo estricto con algunos detalles.

### Reglas fundamentales

- Python distingue mayúsculas de minúsculas: un nombre escrito de dos formas distintas puede representar cosas diferentes.
- La sangría, normalmente de cuatro espacios, define bloques de instrucciones. No se usa solo para decorar.
- Los dos puntos indican que comienza un bloque después de una condición, un bucle, una función o una definición similar.
- Una instrucción suele escribirse por línea; no hace falta terminarla con punto y coma.
- Los comentarios empiezan con el símbolo numeral y sirven para explicar una decisión o dejar una nota para quien lee.
- Los nombres deben ser descriptivos. Por convención, las variables y funciones se escriben en minúsculas separando palabras con guion bajo.


## 2. Variables: nombres para trabajar con datos

Una **variable** es un nombre asociado a una dirección y que esa direccion apunta a un espacio en memoria, puede consultar y, si hace falta, reemplazar durante su ejecución. El nombre permite referirnos a un dato sin tener que repetirlo ni recordar dónde está almacenado.

La forma básica de asignar un valor es:

`nombre = valor`

El signo igual en esta situación se llama **asignación**: significa “guardar el valor de la derecha usando el nombre de la izquierda”. No expresa una igualdad matemática.

### Ideas importantes

- Python crea la variable cuando se ejecuta una asignación; no es necesario declarar por adelantado el tipo.
- Una variable puede recibir otro valor más adelante.
- En una expresión, la variable se reemplaza mentalmente por el valor que contiene.
- El valor de la derecha se calcula antes de guardarse en el nombre de la izquierda.
- Usar una variable antes de asignarle un valor produce un error.

### Nombres válidos y recomendados

Un nombre puede contener letras, números y guion bajo, pero no debe comenzar con un número ni coincidir con una palabra reservada de Python. Conviene evitar nombres de una sola letra, salvo en situaciones muy acotadas, y elegir nombres que expliquen qué representa el dato.

Ejemplos de intención: `edad`, `precio_total`, `nombre_cliente`, `cantidad_de_intentos`.



## 3. Tipos de datos básicos

El **tipo** describe qué clase de valor tenemos y qué operaciones tienen sentido sobre él. Python determina el tipo a partir del valor asignado. Para inspeccionarlo se puede usar la función incorporada `type`.

### Tipos primitivos o simples

| Tipo | Representa | Ejemplos de uso |
| --- | --- | --- |
| `int` | Números enteros | edades, cantidades, puntajes |
| `float` | Números con parte decimal | precios, promedios, medidas |
| `str` | Cadenas de caracteres | nombres, mensajes, códigos |
| `bool` | Valores lógicos: verdadero o falso | respuestas, estados, permisos |
| `NoneType` | Ausencia intencional de valor | dato todavía no disponible |

Los textos se escriben entre comillas simples o dobles. Los valores booleanos se escriben con inicial mayúscula. `None` no es lo mismo que cero, texto vacío o `False`: representa que todavía no hay un valor.

### Conversión de tipos

A veces un dato llega como texto pero necesitamos tratarlo como número. Para convertir se utilizan funciones como `int`, `float`, `str` y `bool`. La conversión solo funciona si el valor tiene un formato compatible; de lo contrario, se produce un error.


````python
mi_int = 5
type(mi_int)
````

````python
my_float = 3.5
type(my_float)
````

````python
mi_str="hola mundo"
type(mi_str)
````

````python
mi_bool=True
type(mi_bool)
````

````python
my_none=None
type(my_none)
"""" Esto no se puede asignar de esta manera"""
````

## 4. Texto: cadenas y operaciones habituales

Una cadena (`str`) es una secuencia de caracteres. Puede contener letras, números, espacios y símbolos; que tenga números escritos no la convierte automáticamente en un número.

Operaciones y recursos para practicar:

- **Concatenar:** unir cadenas.
- **Repetir:** construir una cadena repitiendo otra.
- **Longitud:** obtener cuántos caracteres contiene con `len`.
- **Índice:** acceder a un carácter usando una posición. En Python, las posiciones comienzan en cero.
- **Rebanado:** obtener una parte de la cadena indicando un inicio y un final.
- **Métodos:** operaciones asociadas al texto, como pasar a mayúsculas, quitar espacios externos o separar una cadena.
- **Interpolación:** insertar valores dentro de un texto usando una cadena con formato; es la forma más legible de construir mensajes dinámicos.


## 5. Colecciones: tipos de datos complejos

Cuando necesitamos trabajar con varios datos relacionados, podemos agruparlos en una colección. Las colecciones se diferencian por cómo ordenan, modifican y encuentran sus elementos.

| Colección | Característica principal | Cuándo elegirla |
| --- | --- | --- |
| `list` | Ordenada y modificable | conjunto de elementos que puede crecer o cambiar |
| `tuple` | Ordenada y no modificable | grupo de valores que debe permanecer fijo |
| `dict` | Pares clave–valor | información con atributos identificables |
| `set` | Elementos únicos, sin orden posicional garantizado | eliminar duplicados o consultar pertenencia |

### Listas

Una lista puede contener varios valores, incluso de distintos tipos, aunque normalmente conviene que represente elementos de una misma idea. Se accede a sus elementos por índice, se puede consultar su longitud y se pueden agregar, quitar o reemplazar elementos.

### Tuplas

Una tupla también conserva un orden, pero no está pensada para modificarse. Es útil para representar una agrupación fija, por ejemplo las coordenadas de un punto.

### Diccionarios

Un diccionario relaciona una **clave** con un **valor**. La clave permite acceder a un dato por su significado en lugar de por su posición. Las claves deben ser únicas dentro del diccionario.


## 6. Expresiones y operadores

Una **expresión** combina valores, variables y operadores para producir un resultado. Una instrucción puede guardar ese resultado, mostrarlo o usarlo para tomar una decisión.

### Operadores aritméticos

- Suma, resta, multiplicación y división.
- División entera: obtiene solo la parte entera del cociente.
- Resto o módulo: obtiene lo que sobra de una división. Es muy útil para saber si un número es par.
- Potencia.

Respetar el orden de operaciones: primero paréntesis, luego potencias, multiplicaciones y divisiones, y finalmente sumas y restas. Usar paréntesis cuando mejoren la claridad.

### Comparación

Las comparaciones producen un booleano: igual, distinto, mayor, menor, mayor o igual, menor o igual. Recordar que el operador de comparación de igualdad no es el mismo que el operador de asignación.

### Lógicos

- `and`: todas las condiciones deben ser verdaderas.
- `or`: alcanza con que una condición sea verdadera.
- `not`: invierte un valor lógico.

Los operadores lógicos permiten combinar condiciones, por ejemplo para verificar que una edad esté dentro de un rango. Python evalúa las condiciones de izquierda a derecha y puede dejar de evaluar cuando ya conoce el resultado.


## 8. Entrada y salida

La salida permite comunicar resultados; la función `print` muestra información en pantalla. Conviene que los mensajes sean claros y que indiquen qué representa cada valor.

La función `input` pausa el programa, muestra un mensaje y devuelve lo que se escribió como **texto**. Si esperamos un número, primero debemos convertir esa entrada al tipo adecuado.

### Buenas prácticas

- Explicar qué dato se solicita.
- Convertir la entrada cerca del lugar donde se recibe.
- No asumir que la persona siempre escribirá un valor válido.
- Separar el dato recibido del cálculo que se hará con él.
- Mostrar resultados con unidades, contexto y formato comprensible.


## 9. Condiciones

Una condición permite elegir qué instrucciones ejecutar según una expresión booleana. La estructura comienza con `if`, continúa con una condición y un bloque indentado.

### Estructura conceptual

1. Evaluar una pregunta que pueda responderse como verdadero o falso.
2. Ejecutar el bloque principal si la respuesta es verdadera.
3. Opcionalmente, usar `else` para el caso contrario.
4. Si hay más de dos posibilidades, encadenar alternativas con `elif`.

Cada bloque debe tener una sangría consistente. La condición no necesita compararse explícitamente con `True`: una expresión booleana ya puede usarse como condición.

### Diseñar buenas condiciones

- Formular la pregunta con claridad.
- Cubrir todos los casos relevantes.
- Ordenar las alternativas desde el caso más específico al más general cuando sea necesario.
- Evitar condiciones demasiado largas: dividirlas en variables booleanas con nombres descriptivos.
- Probar los bordes: cero, valores negativos, límites incluidos y datos vacíos.



## 10. Repetición: bucles

Los bucles permiten repetir un conjunto de instrucciones sin copiarlo muchas veces. Una repetición debe tener un criterio claro para avanzar y terminar.

### Bucle `for`

Se usa cuando recorremos una colección o conocemos una secuencia de elementos. En cada vuelta, una variable representa el elemento actual. La función `range` ayuda a generar una secuencia de números; su límite final no se incluye.

### Bucle `while`

Se usa cuando repetimos mientras una condición sea verdadera y no sabemos de antemano cuántas vueltas habrá. Dentro del bucle debe cambiar algo que acerque al programa a la finalización; si no, puede ocurrir un bucle infinito.

### Controlar un bucle

- `break` termina el bucle antes de tiempo cuando se cumple una condición de salida.
- `continue` salta al comienzo de la siguiente vuelta.



## 11. Funciones: dividir y reutilizar soluciones

Una **función** es un bloque de instrucciones con un nombre y una responsabilidad concreta. Permite encapsular una solución, reutilizarla y probarla por separado.

Una función se define con `def`, un nombre, paréntesis para sus parámetros, dos puntos y un bloque indentado. Para usarla hay que invocarla escribiendo su nombre y proporcionando los argumentos necesarios.

### Parámetros, argumentos y retorno

- **Parámetro:** nombre que recibe la función en su definición.
- **Argumento:** valor concreto que se envía al invocarla.
- **`return`:** devuelve un resultado al lugar desde donde se llamó la función.
- Si una función no devuelve explícitamente un valor, su resultado es `None`.

No confundir mostrar un resultado con devolverlo: `print` comunica algo en pantalla; `return` entrega un valor para que otro código pueda guardarlo, compararlo o seguir procesándolo.

### Buen diseño de funciones

- Una función debería hacer una cosa bien definida.
- El nombre debe expresar una acción o una intención.
- Recibir datos por parámetros es preferible a depender de variables externas.
- Evitar funciones demasiado largas.
- Documentar qué recibe y qué devuelve cuando no sea evidente.



## 12. Alcance: dónde existe una variable

El **alcance** o *scope* indica en qué parte del programa se puede consultar un nombre. Una variable creada dentro de una función pertenece normalmente a esa función; fuera de ella no está disponible.

Las variables locales ayudan a que cada función sea independiente y evitan que una parte del programa modifique datos de otra sin querer. Las variables globales pueden parecer cómodas al principio, pero dificultan entender de dónde sale un valor y quién lo cambia.

### Regla práctica

Preferir pasar información como argumento y recibir resultados mediante `return`. Mantener cada dato lo más cerca posible del lugar donde se utiliza.


## 13. Errores y depuración

Programar incluye equivocarse, observar y corregir. La **depuración** es el proceso de encontrar la causa de un comportamiento incorrecto.

### Tipos frecuentes de errores

- **Sintaxis:** Python no puede interpretar la escritura: falta un signo, dos puntos o la sangría es inválida.
- **Nombre:** se usa una variable o función que no existe en ese alcance.
- **Tipo:** se intenta realizar una operación incompatible con el tipo de dato.
- **Índice o clave:** se solicita una posición que no existe o una clave ausente.
- **Lógica:** el programa se ejecuta, pero el resultado no representa lo que queríamos.

### Método de depuración

1. Leer la última línea del mensaje de error: suele indicar la causa.
2. Revisar la línea señalada y también las anteriores.
3. Reproducir el problema con el ejemplo más pequeño posible.
4. Observar los valores intermedios y sus tipos.
5. Formular una hipótesis, cambiar una sola cosa y volver a probar.

La traza de error muestra el camino de llamadas que llevó al problema. No es un castigo: es información para investigar.

## 14. Archivos, módulos y scripts

Un notebook es útil para explorar de forma interactiva. Un **script** es un archivo de texto con extensión `.py` que contiene instrucciones para ejecutar de principio a fin. Ambos usan la misma sintaxis de Python.

### Módulos

Un módulo es un archivo que reúne funciones, variables o clases relacionadas. La instrucción `import` permite reutilizar código de la biblioteca estándar o de otros módulos instalados. Conviene importar solo lo que se necesita y elegir nombres que no oculten nombres incorporados de Python.

### Punto de entrada

En un script es habitual separar las funciones de la parte que inicia el programa. El patrón de punto de entrada permite importar el archivo sin ejecutar automáticamente la parte principal y deja claro dónde comienza el flujo de la aplicación.

### Lectura y escritura de archivos

Para trabajar con archivos, indicar si se abrirá para leer, escribir o agregar contenido, y cerrarlo correctamente al terminar. La forma recomendada utiliza un bloque contextual que garantiza el cierre del archivo incluso si ocurre un error.



## 15. Excepciones y validación

Una **excepción** es una situación anormal durante la ejecución, como convertir un texto no numérico, dividir por cero o no encontrar un archivo. Si no se atiende, el programa se detiene y muestra una traza.

La estructura `try` / `except` permite intentar una operación y definir qué hacer ante ciertos errores. La excepción debe capturarse de forma específica: capturar cualquier error sin analizarlo puede ocultar problemas reales.

### Validar antes de procesar

Siempre que el dato provenga de una persona, un archivo o una red, validar:

- que exista,
- que tenga el tipo esperado,
- que esté dentro del rango permitido,
- que respete el formato necesario.


## 16. Uso de `open` y `with`

La función `open` permite abrir un archivo para leerlo o modificarlo. Recibe la ruta y, opcionalmente, el modo de apertura. Los modos más usados son `r` para leer, `w` para escribir desde cero y `a` para agregar contenido al final. Al trabajar con texto, conviene indicar también la codificación, por ejemplo `encoding="utf-8"`.

Si usamos `open` sin `with`, debemos cerrar el archivo manualmente con `close()`. Si olvidamos hacerlo, el archivo puede quedar abierto y los recursos del sistema no se liberan correctamente:

```python
archivo = open("notas.txt", "r", encoding="utf-8")
contenido = archivo.read()
archivo.close()
```

La forma recomendada es usar `with open(...)`. `with` crea un contexto que cierra el archivo automáticamente al salir del bloque, incluso si dentro ocurre una excepción. Así evitamos olvidos y el código es más seguro y claro:

```python
with open("notas.txt", "r", encoding="utf-8") as archivo:
    contenido = archivo.read()

print(contenido)
```

La diferencia principal es que `open` solo abre el archivo; `with` además administra su ciclo de vida y garantiza que se cierre. Por eso, para la mayoría de los casos, se prefiere `with open(...)`.

Construir un programa que simule una conversación sencilla entre una persona y la computadora. El programa debe reconocer algunas palabras o frases preestablecidas y responder según un conjunto de reglas heurísticas implementadas con condiciones. - Mostrar mensajes claros para las respuestas reconocidas y una respuesta general cuando no se encuentre ninguna coincidencia.
