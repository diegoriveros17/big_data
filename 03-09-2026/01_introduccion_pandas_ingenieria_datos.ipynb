## 1. ¿Qué hace una persona de ingeniería de datos?

Una persona de ingeniería de datos construye y mantiene procesos para que los datos lleguen desde sus **fuentes** hasta las personas o sistemas que los necesitan. Por ejemplo, desde archivos CSV, APIs, sensores o una base de datos hacia un tablero, un modelo de machine learning o un reporte.

Un flujo típico sigue estas etapas:

1. **Extraer:** obtener datos desde una fuente.
2. **Validar y limpiar:** detectar errores, valores faltantes y formatos inconsistentes.
3. **Transformar:** preparar los datos para responder una necesidad del negocio.
4. **Integrar:** combinar fuentes relacionadas.
5. **Publicar:** guardar el resultado para que otros puedan usarlo.

`pandas` es especialmente útil para aprender y prototipar las etapas 2, 3 y 4 con conjuntos de datos que caben en la memoria del equipo.

## 2. ¿Qué es pandas?

`pandas` es una biblioteca de Python orientada al análisis y la manipulación de datos estructurados. Su unidad principal es el **DataFrame**, una tabla con filas y columnas, similar a una hoja de cálculo o al resultado de una consulta SQL.

El alias `pd` es una convención ampliamente usada. Permite reconocer de inmediato que una operación pertenece a pandas.

**Cuándo elegir pandas:** exploración, limpieza, preparación de archivos, automatizaciones pequeñas y pruebas de lógica de transformación. Para datos que no caben en memoria o procesamiento distribuido, se usan herramientas como Spark, pero muchos conceptos se mantienen.

## 3. Estructuras principales: Series y DataFrame

Una **Series** es una sola columna de datos con un índice. Puede representar, por ejemplo, los precios de una tabla o los dias de la semana. 

```python
    dias_semana = ["Lunes", "Martes", "Miercoles", "Jueves", "Viernes", "Sabado", "Domingo"] 
    pd.Series(dias_semana)
```

Un **DataFrame** es una colección de Series que comparten el mismo índice. Cada fila suele representar una observación o evento: una venta, un estudiante, una lectura de sensor o una visita web.

| Concepto | Ejemplo | Equivalencia aproximada |
| --- | --- | --- |
| Fila | Una venta realizada | Registro de una tabla SQL |
| Columna | `fecha_venta` | Campo de una tabla SQL |
| Índice | Identificador interno de la fila | No siempre tiene equivalente directo |
| DataFrame | Tabla de ventas | Tabla o resultado de una consulta |

En proyectos reales, antes de transformar un DataFrame hay que identificar el significado de cada fila, la clave que la identifica y el tipo esperado de cada columna. Esto forma parte del **contrato de datos**.

## 5. Inspección inicial: conocer antes de cambiar

Al recibir un conjunto de datos, primero se explora. Esta revisión permite descubrir columnas inesperadas, tipos erróneos, datos ausentes y posibles problemas de calidad.

| Método o atributo | Para qué se usa |
| --- | --- |
| `df.head()` | Ver las primeras filas y reconocer la estructura rápidamente. |
| `df.tail()` | Revisar las últimas filas; útil para detectar cambios al final de un archivo. |
| `df.shape` | Conocer la cantidad de filas y columnas. |
| `df.columns` | Listar los nombres de las columnas. |
| `df.info()` | Revisar tipos de datos, valores no nulos y uso de memoria. |
| `df.describe()` | Obtener estadísticas descriptivas de columnas numéricas. |
| `df['columna'].value_counts()` | Contar la frecuencia de cada categoría. |

En ingeniería de datos, estas operaciones responden preguntas de calidad: ¿cuántos registros llegaron?, ¿cuántos valores faltan?, ¿la columna fecha realmente es fecha?, ¿apareció una categoría nueva?

## 6. Selección y filtrado: trabajar con la parte necesaria

Seleccionar columnas reduce ruido y hace explícito qué datos necesita una transformación. Filtrar filas permite conservar solo los registros que cumplen una condición.

| Expresión o método | Para qué se usa |
| --- | --- |
| `df['columna']` | Seleccionar una columna como Series. |
| `df[['a', 'b']]` | Seleccionar varias columnas como DataFrame. |
| `df.loc[filas, columnas]` | Seleccionar por etiquetas o condiciones claras. |
| `df.iloc[filas, columnas]` | Seleccionar por posición numérica. |
| `df.query()` | Filtrar con una expresión legible, especialmente en condiciones largas. |

Ejemplo: conservar ventas con importe positivo.

```python
ventas_validas = ventas.loc[ventas['importe'] > 0, ['id_venta', 'fecha', 'importe']]
```

Usar `.loc` hace visible tanto la condición aplicada como las columnas que se conservarán.

## 7. Calidad y limpieza de datos

Los datos reales pueden tener campos vacíos, filas repetidas, errores tipográficos o tipos incorrectos. Limpiar no significa borrar todo lo extraño: significa aplicar una regla de negocio conocida y dejarla documentada.

| Método | Para qué se usa |
| --- | --- |
| `df.isna()` | Identificar valores faltantes. |
| `df.isna().sum()` | Contar valores faltantes por columna. |
| `df.dropna()` | Eliminar filas o columnas con valores faltantes según una regla. |
| `df.fillna()` | Reemplazar valores faltantes por un valor definido. |
| `df.duplicated()` | Detectar filas duplicadas. |
| `df.drop_duplicates()` | Eliminar duplicados, opcionalmente usando columnas clave. |
| `df.rename()` | Normalizar o corregir nombres de columnas. |
| `df.astype()` | Convertir una columna a un tipo específico. |
| `pd.to_datetime()` | Convertir texto u otros valores a fechas. |

Una práctica importante es medir el efecto de la limpieza: comparar `df.shape` antes y después, y registrar cuántos valores o filas se modificaron.

## 8. Transformación: convertir datos en información útil

Transformar consiste en crear, modificar o reorganizar columnas para responder una necesidad concreta. Por ejemplo, derivar el año de una fecha, estandarizar códigos o calcular el importe total de una venta.

| Método o acceso | Para qué se usa |
| --- | --- |
| `df.assign()` | Crear columnas nuevas de forma encadenable. |
| `df.sort_values()` | Ordenar filas por una o más columnas. |
| `df.replace()` | Sustituir valores específicos y normalizar categorías. |
| `df['texto'].str` | Aplicar operaciones de texto: mayúsculas, reemplazos, extracción, etc. |
| `df['fecha'].dt` | Extraer partes de fechas: año, mes, día o día de la semana. |

Se prefieren las operaciones **vectorizadas** de pandas, como sumar columnas directamente, porque trabajan sobre toda la columna y suelen ser más claras y eficientes que recorrer fila por fila.

## 10. Integración de fuentes: combinar tablas

Es frecuente que la información esté separada en distintas tablas. Por ejemplo, podemos tener una tabla con las ventas realizadas y otra con la información de los clientes.

Antes de combinarlas, debemos identificar cuál es la **clave** que relaciona ambas tablas. En este caso, podría ser `id_cliente`.

| Método        | Para qué se usa                                                                            |
| ------------- | ------------------------------------------------------------------------------------------ |
| `pd.merge()`  | Unir columnas de dos DataFrames utilizando una clave, de forma similar a un `JOIN` de SQL. |
| `df.join()`   | Unir DataFrames principalmente utilizando sus índices.                                     |
| `pd.concat()` | Combinar DataFrames apilándolos por filas o columnas.                                      |

---

### `pd.merge()`

Supongamos que tenemos estos dos DataFrames:

Visualmente:

**clientes**

| id_cliente | nombre |
| ---------- | ------ |
| 1          | Ana    |
| 2          | Carlos |
| 3          | Lucía  |

**ventas**

| id_venta | id_cliente | total |
| -------- | ---------- | ----- |
| 101      | 1          | 15000 |
| 102      | 2          | 22000 |
| 103      | 4          | 18000 |

Ambas tablas tienen la columna `id_cliente`, por lo tanto podemos utilizarla como clave para relacionarlas.

### `inner`

Conserva únicamente los clientes que aparecen en ambas tablas.


```python
resultado = pd.merge(
    ventas,
    clientes,
    on="id_cliente",
    how="inner"
)

```

Resultado:

```text
   id_venta  id_cliente  total  nombre
0       101           1  15000     Ana
1       102           2  22000  Carlos
```

El cliente `3` no aparece porque no tiene ventas, mientras que el cliente `4` tampoco aparece porque no existe en la tabla `clientes`.

---

### `left`

Conserva todas las filas del DataFrame ubicado a la izquierda.


```python
resultado = pd.merge(
    ventas,
    clientes,
    on="id_cliente",
    how="left"
)
```

Resultado:

```text
   id_venta  id_cliente  total  nombre
0       101           1  15000     Ana
1       102           2  22000  Carlos
2       103           4  18000     NaN
```

Como `ventas` está a la izquierda, se conservan todas sus filas.

El `id_cliente = 4` no tiene información asociada en la tabla `clientes`, por eso aparece `NaN` en `nombre`.

Este tipo de unión es muy útil cuando queremos conservar todos nuestros registros principales aunque no encontremos información relacionada en otra tabla.

---

### `right`

Conserva todas las filas del DataFrame ubicado a la derecha.


```python
resultado = pd.merge(
    ventas,
    clientes,
    on="id_cliente",
    how="right"
)
```

## 11. Escritura de resultados: la etapa de publicación

Cuando una transformación está validada, el resultado se guarda en un formato que pueda consumir el siguiente paso del proceso.

| Método | Para qué se usa |
| --- | --- |
| `df.to_csv()` | Exportar una tabla a CSV, práctico para interoperabilidad. |
| `df.to_excel()` | Generar un archivo Excel para usuarios de negocio. |
| `df.to_parquet()` | Guardar datos tabulares de forma eficiente, conservando tipos. |
| `df.to_sql()` | Escribir datos en una tabla de base de datos. |

Ejemplo conceptual:

```python
resumen.to_parquet('resumen_ventas.parquet', index=False)
```

El parámetro `index=False` evita guardar el índice técnico como una columna adicional cuando no forma parte del dato de negocio.

## 12. Buenas prácticas para empezar

1. **Inspeccionar antes de transformar.** Revisar estructura, tipos y faltantes antes de escribir reglas.
2. **No modificar la fuente original.** Conservar los datos de entrada y producir una salida nueva.
3. **Usar nombres claros.** Preferir `ventas_limpias` a `df2`.
4. **Validar cada etapa.** Controlar filas, columnas, nulos y duplicados después de operaciones importantes.
5. **Documentar supuestos.** Explicar por qué se eliminaron filas o cómo se rellenaron faltantes.
6. **Elegir tipos correctos.** Fechas como fechas, importes como números y categorías como texto o tipo categórico.

### Idea clave

Aprender pandas no consiste solo en manipular tablas. Consiste en desarrollar criterio para entender la calidad, el significado y el recorrido de los datos. Ese criterio es la base para trabajar después con bases de datos, procesos ETL/ELT y plataformas de Big Data.
